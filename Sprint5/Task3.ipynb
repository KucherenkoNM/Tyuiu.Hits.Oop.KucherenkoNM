В каждом из классов должны быть реализованы новые атрибуты и методы, а также переопределены некоторые методы базового класса для демонстрации полиморфизма.


#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) исользуйтие в проекте коллекции, делегаты, события.

```
using System;
using System.Collections.Generic;
using System.Linq;

public interface ILogistics
{
    void ArrangeDelivery();
}

public interface ICertified
{
    void ShowCertificate();
}

class Supplier
{
    private int supplierId;
    private string companyName;
    private string productType;
    private double rating;

    public string ContactEmail { get; set; }
    public string ContractNumber { get; set; }

    public event Action<string> RatingChanged;

    public void ShowContract()
    {
        Console.WriteLine($"Номер договора с {CompanyName}: {ContractNumber}");
    }

    public void ShowContactInfo()
    {
        Console.WriteLine($"Контактный email компании {CompanyName}: {ContactEmail}");
    }

    public void GetCompanyInfo(bool showDetails)
    {
        GetCompanyInfo();
        if (showDetails)
        {
            ShowRating();
            ShowContactInfo();
        }
    }

    public int SupplierId
    {
        get { return supplierId; }
        set { supplierId = value; }
    }

    public string CompanyName
    {
        get { return companyName; }
        set { companyName = value; }
    }

    public string ProductType
    {
        get { return productType; }
        set { productType = value; }
    }

    public double Rating
    {
        get { return rating; }
        set { rating = value; }
    }

    public Supplier(
        int supplierId,
        string companyName,
        string productType,
        double rating)
    {
        SupplierId = supplierId;
        CompanyName = companyName;
        ProductType = productType;
        Rating = rating;
    }

    public virtual void GetCompanyInfo()
    {
        Console.WriteLine(
            $"ID поставщика: {SupplierId}, компания: {CompanyName}, " +
            $"тип продукции: {ProductType}, рейтинг: {Rating}");
    }

    public virtual void ProvideQuote()
    {
        Console.WriteLine(
            $"Компания {CompanyName} предоставляет котировку на продукцию: {ProductType}");
    }

    public virtual void SubmitOrder()
    {
        Console.WriteLine($"Заказ отправлен поставщику {CompanyName}");
    }

    public void ShowRating()
    {
        Console.WriteLine($"Рейтинг компании {CompanyName}: {Rating}");
    }

    public void UpdateRating(double newRating)
    {
        Rating = newRating;
        RatingChanged?.Invoke($"Рейтинг компании {CompanyName} изменён на {newRating}");
    }

    public string GetDescription()
    {
        return $"{CompanyName} — {ProductType}, рейтинг: {Rating}";
    }
}

class Manufacturer : Supplier
{
    private int foundedYear;
    private int productionCapacity;
    private string factoryLocation;

    public int EmployeeCount { get; set; }
    public int EquipmentCount { get; set; }

    public void ShowEquipmentCount()
    {
        Console.WriteLine($"Количество оборудования {CompanyName}: {EquipmentCount}");
    }

    public void ShowEmployeeCount()
    {
        Console.WriteLine($"Количество сотрудников {CompanyName}: {EmployeeCount}");
    }

    public int FoundedYear
    {
        get { return foundedYear; }
        set { foundedYear = value; }
    }

    public int ProductionCapacity
    {
        get { return productionCapacity; }
        set { productionCapacity = value; }
    }

    public string FactoryLocation
    {
        get { return factoryLocation; }
        set { factoryLocation = value; }
    }

    public Manufacturer(
        int supplierId,
        string companyName,
        string productType,
        double rating,
        int foundedYear,
        int productionCapacity,
        string factoryLocation)
        : base(supplierId, companyName, productType, rating)
    {
        FoundedYear = foundedYear;
        ProductionCapacity = productionCapacity;
        FactoryLocation = factoryLocation;
    }

    public override void ProvideQuote()
    {
        Console.WriteLine(
            $"Компания {CompanyName} предоставляет котировку на продукцию: {ProductType}. " +
            $"Год основания: {FoundedYear}");
    }

    public void StartProduction()
    {
        Console.WriteLine($"Производитель {CompanyName} запускает производство.");
    }

    public void CheckCapacity()
    {
        Console.WriteLine(
            $"Производственная мощность {CompanyName}: {ProductionCapacity} единиц.");
    }

    public void ShowFactoryLocation()
    {
        Console.WriteLine(
            $"Производство компании {CompanyName} находится: {FactoryLocation}");
    }
}

class ElectronicsManufacturer : Manufacturer
{
    private int warrantyYears;
    private bool hasResearchCenter;
    private string deviceCategory;

    public int WarrantyYears
    {
        get { return warrantyYears; }
        set { warrantyYears = value; }
    }

    public bool HasResearchCenter
    {
        get { return hasResearchCenter; }
        set { hasResearchCenter = value; }
    }

    public string DeviceCategory
    {
        get { return deviceCategory; }
        set { deviceCategory = value; }
    }

    public ElectronicsManufacturer(
        int supplierId,
        string companyName,
        string productType,
        double rating,
        int foundedYear,
        int productionCapacity,
        string factoryLocation,
        int warrantyYears,
        bool hasResearchCenter,
        string deviceCategory)
        : base(
            supplierId,
            companyName,
            productType,
            rating,
            foundedYear,
            productionCapacity,
            factoryLocation)
    {
        WarrantyYears = warrantyYears;
        HasResearchCenter = hasResearchCenter;
        DeviceCategory = deviceCategory;
    }

    public void AssembleDevice()
    {
        Console.WriteLine(
            $"{CompanyName} собирает устройства категории: {DeviceCategory}");
    }

    public void TestProducts()
    {
        Console.WriteLine($"{CompanyName} проводит тестирование электроники.");
    }

    public void ShowWarranty()
    {
        Console.WriteLine(
            $"Гарантия на продукцию {CompanyName}: {WarrantyYears} года.");
    }
}

class Retailer : Supplier
{
    private string storeLocation;
    private int storeCount;
    private bool deliveryAvailable;

    public bool OnlineStoreAvailable { get; set; }
    public double DiscountPercent { get; set; }

    public void ShowDiscount()
    {
        Console.WriteLine($"Скидка в {CompanyName}: {DiscountPercent}%");
    }

    public void ShowOnlineStoreInfo()
    {
        Console.WriteLine(
            $"Интернет-магазин {CompanyName}: {(OnlineStoreAvailable ? "есть" : "нет")}");
    }

    public string StoreLocation
    {
        get { return storeLocation; }
        set { storeLocation = value; }
    }

    public int StoreCount
    {
        get { return storeCount; }
        set { storeCount = value; }
    }

    public bool DeliveryAvailable
    {
        get { return deliveryAvailable; }
        set { deliveryAvailable = value; }
    }

    public Retailer(
        int supplierId,
        string companyName,
        string productType,
        double rating,
        string storeLocation,
        int storeCount,
        bool deliveryAvailable)
        : base(supplierId, companyName, productType, rating)
    {
        StoreLocation = storeLocation;
        StoreCount = storeCount;
        DeliveryAvailable = deliveryAvailable;
    }

    public override void SubmitOrder()
    {
        Console.WriteLine(
            $"Заказ отправлен ритейлеру {CompanyName}. " +
            $"Расположение магазина: {StoreLocation}");
    }

    public void ShowStoreLocation()
    {
        Console.WriteLine(
            $"Магазин {CompanyName} расположен: {StoreLocation}");
    }

    public void RequestQuote(Supplier supplier)
    {
        Console.WriteLine(
            $"Ритейлер {CompanyName} запрашивает котировку у {supplier.CompanyName}");

        supplier.ProvideQuote();
    }

    public void ShowDeliveryInfo()
    {
        Console.WriteLine(
            $"Доставка у {CompanyName}: {(DeliveryAvailable ? "доступна" : "недоступна")}");
    }
}

class Importer : Supplier, ILogistics, ICertified
{
    private string originCountry;
    private int deliveryDays;
    private string transportType;

    public int CustomsClearanceDays { get; set; }

    public void ShowCustomsInfo()
    {
        Console.WriteLine(
            $"Таможенное оформление {CompanyName}: {CustomsClearanceDays} дней.");
    }

    public string OriginCountry
    {
        get { return originCountry; }
        set { originCountry = value; }
    }

    public int DeliveryDays
    {
        get { return deliveryDays; }
        set { deliveryDays = value; }
    }

    public string TransportType
    {
        get { return transportType; }
        set { transportType = value; }
    }

    public Importer(
        int supplierId,
        string companyName,
        string productType,
        double rating,
        string originCountry,
        int deliveryDays,
        string transportType)
        : base(supplierId, companyName, productType, rating)
    {
        OriginCountry = originCountry;
        DeliveryDays = deliveryDays;
        TransportType = transportType;
    }

    public override void GetCompanyInfo()
    {
        Console.WriteLine(
            $"ID поставщика: {SupplierId}, компания: {CompanyName}, " +
            $"тип продукции: {ProductType}, страна происхождения: {OriginCountry}");
    }

    public void ImportGoods()
    {
        Console.WriteLine(
            $"Импортер {CompanyName} поставляет продукцию из страны: {OriginCountry}");
    }

    void ILogistics.ArrangeDelivery()
    {
        Console.WriteLine(
            $"{CompanyName} организует доставку. " +
            $"Транспорт: {TransportType}, срок: {DeliveryDays} дней.");
    }

    void ICertified.ShowCertificate()
    {
        Console.WriteLine(
            $"Импортер {CompanyName} имеет сертификат на ввоз продукции.");
    }
}

class SupplierService
{
    private readonly Supplier _supplier;

    public SupplierService(Supplier supplier)
    {
        _supplier = supplier;
    }

    public void ShowSupplier()
    {
        _supplier.GetCompanyInfo();
    }
}

class SupplierCollection<T> where T : Supplier
{
    private List<T> _suppliers = new List<T>();

    public void Add(T supplier)
    {
        _suppliers.Add(supplier);
    }

    public void Remove(T supplier)
    {
        _suppliers.Remove(supplier);
    }

    public void DisplaySuppliers()
    {
        foreach (var supplier in _suppliers)
        {
            supplier.GetCompanyInfo();
        }
    }
}

Manufacturer manufacturer = new Manufacturer(
    1, "TechnoFactory", "Оборудование", 4.7, 1998, 5000, "Москва");

ElectronicsManufacturer electronicsManufacturer = new ElectronicsManufacturer(
    2, "ElectroTech", "Электроника", 4.9, 2005, 10000,
    "Санкт-Петербург", 3, true, "Смартфоны");

Retailer retailer = new Retailer(
    3, "CityStore", "Бытовая техника", 4.5, "Москва", 12, true);

Importer importer = new Importer(
    4, "GlobalImport", "Кофе", 4.8, "Бразилия", 14, "Морской транспорт");

manufacturer.EmployeeCount = 350;
manufacturer.ContractNumber = "CTR-001";
manufacturer.EquipmentCount = 120;
manufacturer.ContactEmail = "info@technofactory.example";

electronicsManufacturer.ContactEmail = "info@electrotech.example";

retailer.OnlineStoreAvailable = true;
retailer.DiscountPercent = 10;
retailer.ContactEmail = "support@citystore.example";

importer.CustomsClearanceDays = 5;
importer.ContactEmail = "contact@globalimport.example";

List<Supplier> suppliers = new List<Supplier>
{
    manufacturer,
    electronicsManufacturer,
    retailer,
    importer
};

Console.WriteLine("=== СПИСОК ПОСТАВЩИКОВ ===");

foreach (Supplier supplier in suppliers)
{
    Console.WriteLine(supplier.GetDescription());
}

Console.WriteLine("\n=== ПОИСК ПОСТАВЩИКА ===");

Dictionary<int, Supplier> supplierDictionary = new Dictionary<int, Supplier>();

foreach (Supplier supplier in suppliers)
{
    supplierDictionary.Add(supplier.SupplierId, supplier);
}

if (supplierDictionary.TryGetValue(2, out Supplier foundSupplier))
{
    Console.WriteLine($"Найден: {foundSupplier.GetDescription()}");
}
else
{
    Console.WriteLine("Поставщик не найден.");
}

Console.WriteLine("\n=== ПРОВЕРКА ДУБЛИКАТОВ ===");

HashSet<int> supplierIds = new HashSet<int>();

foreach (Supplier supplier in suppliers)
{
    supplierIds.Add(supplier.SupplierId);
}

bool isAdded = supplierIds.Add(2);

Console.WriteLine($"Повторный ID добавлен: {isAdded}");
Console.WriteLine($"Уникальных поставщиков: {supplierIds.Count}");

Console.WriteLine("\n=== ДЕЛЕГАТ ===");

Func<Supplier, string> descriptionDelegate = s => s.GetDescription();

Console.WriteLine(descriptionDelegate(manufacturer));

Console.WriteLine("\n=== ЛЯМБДА-ВЫРАЖЕНИЕ ===");

var topSuppliers = suppliers.Where(s => s.Rating >= 4.7);

foreach (Supplier supplier in topSuppliers)
{
    Console.WriteLine(supplier.GetDescription());
}

Console.WriteLine("\n=== СОБЫТИЕ ===");

retailer.RatingChanged += message => Console.WriteLine(message);

retailer.UpdateRating(4.7);
```
