### Вариант задания № 14


<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Создать базовый класс `Supplier` в C#, который будет представлять информацию о поставщиках товаров или услуг. На основе этого класса разработать 2–3 производных класса, демонстрирующих принципы наследования и полиморфизма. В каждом из классов должны быть реализованы новые атрибуты и методы, а также переопределены некоторые методы базового класса для демонстрации полиморфизма.

<h3 style="color:DodgerBlue">Требования к базовому классу Supplier:</h3>

----

#### Атрибуты

| Атрибут | Назначение |
|---|---|
| `SupplierId` | ID поставщика |
| `CompanyName` | Название компании |
| `ProductType` | Тип продукции |

#### Методы

| Метод | Назначение |
|---|---|
| `GetCompanyInfo()` | Получение информации о компании |
| `ProvideQuote()` | Предоставление котировки на товары или услуги |
| `SubmitOrder()` | Отправка заказа поставщику |

<h3 style="color:DodgerBlue">Требования к производным классам:</h3>

----

1. **Производитель (`Manufacturer`)**  
   Дополнительный атрибут: `FoundedYear` — год основания компании.  
   Метод `ProvideQuote()` должен быть переопределен для включения информации о годе основания компании в котировку.

2. **Ритейлер (`Retailer`)**  
   Дополнительный атрибут: `StoreLocation` — расположение магазина.  
   Метод `SubmitOrder()` должен быть переопределен для добавления информации о расположении магазина при отправке заказа.

3. **Импортер (`Importer`)**  
   Дополнительный атрибут: `OriginCountry` — страна происхождения товара.  
   Метод `GetCompanyInfo()` должен быть переопределен для отображения страны происхождения товара вместе с остальной информацией о компании.


<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
class Supplier
{
    private int supplierId;
    private string companyName;
    private string productType;

    public int SupplierId
    {
        get { return supplierId; }
        set { supplierId = value; }
    }

    public string CompanyName
    {
        get { return companyName; }
        set { companyName = value; }
    }

    public string ProductType
    {
        get { return productType; }
        set { productType = value; }
    }

    public Supplier(int supplierId, string companyName, string productType)
    {
        SupplierId = supplierId;
        CompanyName = companyName;
        ProductType = productType;
    }

    public virtual void GetCompanyInfo()
    {
        Console.WriteLine(
            $"ID поставщика: {SupplierId}, компания: {CompanyName}, тип продукции: {ProductType}");
    }

    public virtual void ProvideQuote()
    {
        Console.WriteLine(
            $"Компания {CompanyName} предоставляет котировку на продукцию: {ProductType}");
    }

    public virtual void SubmitOrder()
    {
        Console.WriteLine($"Заказ отправлен поставщику {CompanyName}");
    }
}

class Manufacturer : Supplier
{
    private int foundedYear;

    public int FoundedYear
    {
        get { return foundedYear; }
        set { foundedYear = value; }
    }

    public Manufacturer(
        int supplierId,
        string companyName,
        string productType,
        int foundedYear)
        : base(supplierId, companyName, productType)
    {
        FoundedYear = foundedYear;
    }

    public override void ProvideQuote()
    {
        Console.WriteLine(
            $"Компания {CompanyName} предоставляет котировку на продукцию: {ProductType}. " +
            $"Год основания компании: {FoundedYear}");
    }

    public void StartProduction()
    {
        Console.WriteLine($"Производитель {CompanyName} запускает производство");
    }
}

class Retailer : Supplier
{
    private string storeLocation;

    public string StoreLocation
    {
        get { return storeLocation; }
        set { storeLocation = value; }
    }

    public Retailer(
        int supplierId,
        string companyName,
        string productType,
        string storeLocation)
        : base(supplierId, companyName, productType)
    {
        StoreLocation = storeLocation;
    }

    public override void SubmitOrder()
    {
        Console.WriteLine(
            $"Заказ отправлен ритейлеру {CompanyName}. Расположение магазина: {StoreLocation}");
    }

    public void ShowStoreLocation()
    {
        Console.WriteLine($"Магазин {CompanyName} расположен: {StoreLocation}");
    }
}

class Importer : Supplier
{
    private string originCountry;

    public string OriginCountry
    {
        get { return originCountry; }
        set { originCountry = value; }
    }

    public Importer(
        int supplierId,
        string companyName,
        string productType,
        string originCountry)
        : base(supplierId, companyName, productType)
    {
        OriginCountry = originCountry;
    }

    public override void GetCompanyInfo()
    {
        Console.WriteLine(
            $"ID поставщика: {SupplierId}, компания: {CompanyName}, тип продукции: {ProductType}, " +
            $"страна происхождения товара: {OriginCountry}");
    }

    public void ImportGoods()
    {
        Console.WriteLine(
            $"Импортер {CompanyName} поставляет продукцию из страны: {OriginCountry}");
    }
}

Manufacturer manufacturer = new Manufacturer(
    1, "TechnoFactory", "Электроника", 1998);

Retailer retailer = new Retailer(
    2, "CityStore", "Бытовая техника", "Москва");

Importer importer = new Importer(
    3, "GlobalImport", "Кофе", "Бразилия");

Supplier supplier1 = manufacturer;
Supplier supplier2 = retailer;
Supplier supplier3 = importer;

Console.WriteLine("=== ПРОИЗВОДИТЕЛЬ ===");
supplier1.GetCompanyInfo();
supplier1.ProvideQuote();
supplier1.SubmitOrder();
manufacturer.StartProduction();

Console.WriteLine();

Console.WriteLine("=== РИТЕЙЛЕР ===");
supplier2.GetCompanyInfo();
supplier2.ProvideQuote();
supplier2.SubmitOrder();
retailer.ShowStoreLocation();

Console.WriteLine();

Console.WriteLine("=== ИМПОРТЕР ===");
supplier3.GetCompanyInfo();
supplier3.ProvideQuote();
supplier3.SubmitOrder();
importer.ImportGoods();
```
