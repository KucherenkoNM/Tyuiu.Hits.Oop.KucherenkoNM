### Вариант задания №14


<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Создать базовый класс `Supplier` в C#, который будет представлять информацию о поставщиках товаров или услуг. На основе этого класса разработать 2–3 производных класса, демонстрирующих принципы наследования и полиморфизма. В каждом из классов должны быть реализованы новые атрибуты и методы, а также переопределены некоторые методы базового класса для демонстрации полиморфизма.

#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) и реализуйте простое, сложное и множественное наследование

<h2 style="color:DodgerBlue">Реализация:</h2>

```
using System;

public interface ILogistics
{
    void ArrangeDelivery();
}

public interface ICertified
{
    void ShowCertificate();
}

class Supplier
{
    private int supplierId;
    private string companyName;
    private string productType;
    private double rating;

    public int SupplierId
    {
        get { return supplierId; }
        set { supplierId = value; }
    }

    public string CompanyName
    {
        get { return companyName; }
        set { companyName = value; }
    }

    public string ProductType
    {
        get { return productType; }
        set { productType = value; }
    }

    public double Rating
    {
        get { return rating; }
        set { rating = value; }
    }

    public Supplier(
        int supplierId,
        string companyName,
        string productType,
        double rating)
    {
        SupplierId = supplierId;
        CompanyName = companyName;
        ProductType = productType;
        Rating = rating;
    }

    public virtual void GetCompanyInfo()
    {
        Console.WriteLine(
            $"ID поставщика: {SupplierId}, компания: {CompanyName}, " +
            $"тип продукции: {ProductType}, рейтинг: {Rating}");
    }

    public virtual void ProvideQuote()
    {
        Console.WriteLine(
            $"Компания {CompanyName} предоставляет котировку на продукцию: {ProductType}");
    }

    public virtual void SubmitOrder()
    {
        Console.WriteLine($"Заказ отправлен поставщику {CompanyName}");
    }

    public void ShowRating()
    {
        Console.WriteLine($"Рейтинг компании {CompanyName}: {Rating}");
    }
}

class Manufacturer : Supplier
{
    private int foundedYear;
    private int productionCapacity;
    private string factoryLocation;

    public int FoundedYear
    {
        get { return foundedYear; }
        set { foundedYear = value; }
    }

    public int ProductionCapacity
    {
        get { return productionCapacity; }
        set { productionCapacity = value; }
    }

    public string FactoryLocation
    {
        get { return factoryLocation; }
        set { factoryLocation = value; }
    }

    public Manufacturer(
        int supplierId,
        string companyName,
        string productType,
        double rating,
        int foundedYear,
        int productionCapacity,
        string factoryLocation)
        : base(supplierId, companyName, productType, rating)
    {
        FoundedYear = foundedYear;
        ProductionCapacity = productionCapacity;
        FactoryLocation = factoryLocation;
    }

    public override void ProvideQuote()
    {
        Console.WriteLine(
            $"Компания {CompanyName} предоставляет котировку на продукцию: {ProductType}. " +
            $"Год основания: {FoundedYear}");
    }

    public void StartProduction()
    {
        Console.WriteLine($"Производитель {CompanyName} запускает производство.");
    }

    public void CheckCapacity()
    {
        Console.WriteLine(
            $"Производственная мощность {CompanyName}: {ProductionCapacity} единиц.");
    }

    public void ShowFactoryLocation()
    {
        Console.WriteLine(
            $"Производство компании {CompanyName} находится: {FactoryLocation}");
    }
}

class ElectronicsManufacturer : Manufacturer
{
    private int warrantyYears;
    private bool hasResearchCenter;
    private string deviceCategory;

    public int WarrantyYears
    {
        get { return warrantyYears; }
        set { warrantyYears = value; }
    }

    public bool HasResearchCenter
    {
        get { return hasResearchCenter; }
        set { hasResearchCenter = value; }
    }

    public string DeviceCategory
    {
        get { return deviceCategory; }
        set { deviceCategory = value; }
    }

    public ElectronicsManufacturer(
        int supplierId,
        string companyName,
        string productType,
        double rating,
        int foundedYear,
        int productionCapacity,
        string factoryLocation,
        int warrantyYears,
        bool hasResearchCenter,
        string deviceCategory)
        : base(
            supplierId,
            companyName,
            productType,
            rating,
            foundedYear,
            productionCapacity,
            factoryLocation)
    {
        WarrantyYears = warrantyYears;
        HasResearchCenter = hasResearchCenter;
        DeviceCategory = deviceCategory;
    }

    public void AssembleDevice()
    {
        Console.WriteLine(
            $"{CompanyName} собирает устройства категории: {DeviceCategory}");
    }

    public void TestProducts()
    {
        Console.WriteLine($"{CompanyName} проводит тестирование электроники.");
    }

    public void ShowWarranty()
    {
        Console.WriteLine(
            $"Гарантия на продукцию {CompanyName}: {WarrantyYears} года.");
    }
}

class Retailer : Supplier
{
    private string storeLocation;
    private int storeCount;
    private bool deliveryAvailable;

    public string StoreLocation
    {
        get { return storeLocation; }
        set { storeLocation = value; }
    }

    public int StoreCount
    {
        get { return storeCount; }
        set { storeCount = value; }
    }

    public bool DeliveryAvailable
    {
        get { return deliveryAvailable; }
        set { deliveryAvailable = value; }
    }

    public Retailer(
        int supplierId,
        string companyName,
        string productType,
        double rating,
        string storeLocation,
        int storeCount,
        bool deliveryAvailable)
        : base(supplierId, companyName, productType, rating)
    {
        StoreLocation = storeLocation;
        StoreCount = storeCount;
        DeliveryAvailable = deliveryAvailable;
    }

    public override void SubmitOrder()
    {
        Console.WriteLine(
            $"Заказ отправлен ритейлеру {CompanyName}. " +
            $"Расположение магазина: {StoreLocation}");
    }

    public void ShowStoreLocation()
    {
        Console.WriteLine(
            $"Магазин {CompanyName} расположен: {StoreLocation}");
    }

    public void RequestQuote(Supplier supplier)
    {
        Console.WriteLine(
            $"Ритейлер {CompanyName} запрашивает котировку у {supplier.CompanyName}");

        supplier.ProvideQuote();
    }

    public void ShowDeliveryInfo()
    {
        Console.WriteLine(
            $"Доставка у {CompanyName}: {(DeliveryAvailable ? "доступна" : "недоступна")}");
    }
}

class Importer : Supplier, ILogistics, ICertified
{
    private string originCountry;
    private int deliveryDays;
    private string transportType;

    public string OriginCountry
    {
        get { return originCountry; }
        set { originCountry = value; }
    }

    public int DeliveryDays
    {
        get { return deliveryDays; }
        set { deliveryDays = value; }
    }

    public string TransportType
    {
        get { return transportType; }
        set { transportType = value; }
    }

    public Importer(
        int supplierId,
        string companyName,
        string productType,
        double rating,
        string originCountry,
        int deliveryDays,
        string transportType)
        : base(supplierId, companyName, productType, rating)
    {
        OriginCountry = originCountry;
        DeliveryDays = deliveryDays;
        TransportType = transportType;
    }

    public override void GetCompanyInfo()
    {
        Console.WriteLine(
            $"ID поставщика: {SupplierId}, компания: {CompanyName}, " +
            $"тип продукции: {ProductType}, страна происхождения: {OriginCountry}");
    }

    public void ImportGoods()
    {
        Console.WriteLine(
            $"Импортер {CompanyName} поставляет продукцию из страны: {OriginCountry}");
    }

    public void ArrangeDelivery()
    {
        Console.WriteLine(
            $"{CompanyName} организует доставку. " +
            $"Транспорт: {TransportType}, срок: {DeliveryDays} дней.");
    }

    public void ShowCertificate()
    {
        Console.WriteLine(
            $"Импортер {CompanyName} имеет сертификат на ввоз продукции.");
    }
}

Manufacturer manufacturer = new Manufacturer(
    1,
    "TechnoFactory",
    "Оборудование",
    4.7,
    1998,
    5000,
    "Москва");

ElectronicsManufacturer electronicsManufacturer =
    new ElectronicsManufacturer(
        2,
        "ElectroTech",
        "Электроника",
        4.9,
        2005,
        10000,
        "Санкт-Петербург",
        3,
        true,
        "Смартфоны");

Retailer retailer = new Retailer(
    3,
    "CityStore",
    "Бытовая техника",
    4.5,
    "Москва",
    12,
    true);

Importer importer = new Importer(
    4,
    "GlobalImport",
    "Кофе",
    4.8,
    "Бразилия",
    14,
    "Морской транспорт");

Console.WriteLine("=== ПРОИЗВОДИТЕЛЬ ===");
manufacturer.GetCompanyInfo();
manufacturer.ProvideQuote();
manufacturer.StartProduction();
manufacturer.CheckCapacity();
manufacturer.ShowFactoryLocation();

Console.WriteLine();

Console.WriteLine("=== ПРОИЗВОДИТЕЛЬ ЭЛЕКТРОНИКИ ===");
electronicsManufacturer.GetCompanyInfo();
electronicsManufacturer.ProvideQuote();
electronicsManufacturer.StartProduction();
electronicsManufacturer.AssembleDevice();
electronicsManufacturer.TestProducts();
electronicsManufacturer.ShowWarranty();

Console.WriteLine();

Console.WriteLine("=== РИТЕЙЛЕР ===");
retailer.GetCompanyInfo();
retailer.SubmitOrder();
retailer.ShowStoreLocation();
retailer.ShowDeliveryInfo();

Console.WriteLine();

Console.WriteLine("=== ИМПОРТЕР ===");
importer.GetCompanyInfo();
importer.ImportGoods();
importer.ArrangeDelivery();
importer.ShowCertificate();

Console.WriteLine();

Console.WriteLine("=== ВЗАИМОДЕЙСТВИЕ ОБЪЕКТОВ ===");
retailer.RequestQuote(manufacturer);
retailer.RequestQuote(importer);
```
